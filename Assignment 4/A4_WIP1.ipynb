### Part (iii) [1pt EXPLORATORY]

The following helper code will be used to process the data before we can train our LSTM model. In point form comment on what processing steps are performed in the code provided below and why these steps are necessary or beneficial to training and LSTM.

````python
def preprocess_string(str1):
    # remove all non-word characters excluding number and letters
    str1= re.sub(r"[^\w\s]",'',str1)
    # remove all whitespace with no space
    str1= re.sub(r"\s",'',str1)
    # replace digits with no space
    str1= re.sub(r"\d",'',str1)
    return str1

def preprocess_sentence(sen1):
    word_list=[]
    stop_word = set(stopwords.words("english"))
    for word in sen1.lower().split():
        word = preprocess_string(word)
        if word not in stop_word and word!='':
            word_list.append(word)
    return word_list

def get_stoi(data):
    word_list=[]
    for review in data:
        word_list.extend(preprocess_sentence(review))
    corpus = Counter(word_list)
    print(corpus.get)
    # sorting on the basis of most common words
    corpus_ =sorted(corpus,key= corpus.get,reverse=True)[:1000]
    # creating a dict
    stoi =  {ch:i+1 for i,ch in enumerate(corpus_)}
    return stoi

def tokenize(data, labels, stoi):
    # tokenize
    data_encoded = []
    for review in data:
        data_encoded.append([stoi[word] for word in preprocess_sentence(review)
                             if word in stoi.keys()])

    labels_encoded = [1 if label =='positive' else 0 for label in labels]

    return np.array(data_encoded, dtype=object), np.array(labels_encoded)

def padding_(sentences, seq_len):
    features = np.zeros((len(sentences), seq_len),dtype=int)
    for ii, review in enumerate(sentences):
        if len(review)!=0:
            features[ii, -len(review):] = np.array(review)[:seq_len]

    return features
````

````python
'''
PROVIDE YOUR ANSWER BELOW

==== preprocess_string – clean each token ====

  * Removes punctuation, whitespace, and digits → keeps only letters.
  * Reduces noise and vocabulary size, so the LSTM focuses on meaningful word patterns.

==== preprocess_sentence – normalize and remove stopwords ====

  * Lowercases, splits into words, applies `preprocess_string`, then drops English stopwords and empty tokens.
  * Emphasizes content words (e.g., “great”, “boring”) that carry sentiment, improving signal for the LSTM.

==== get_stoi – build a limited vocabulary ====

  * Collects all cleaned words from the reviews, counts them with `Counter`, keeps the **top 1000** most frequent, and assigns each word an index starting at 1.
  * Limits vocabulary size → fewer parameters, less overfitting, faster training, and avoids extremely rare words.

==== tokenize – convert text to integer sequences + encode labels ====

  * For each review, replaces each word (if in `stoi`) with its integer id → produces variable-length integer lists.
  * Encodes labels as `1` for positive and `0` for negative → needed for numeric loss functions.
  * LSTM can only take numeric sequences, not raw strings.

==== padding_ – pad / truncate to fixed length ====

  * Creates a 2D array `[num_reviews, seq_len]` filled with zeros and right-aligns each review sequence (cuts off if longer than `seq_len`).
  * Gives all inputs the same length so we can form batches and feed them as a tensor into the LSTM; zeros naturally act as `<PAD>` tokens.






'''
````

````python
SEED = 42

# 1. Split Data
def data_split(data, y, train_split, val_split):
    x = np.array(data)
    y = np.array(y)
    n = len(data)

    rng = np.random.default_rng(SEED)
    idx = rng.permutation(n)

    n_train = int(train_split * n)
    n_valid = int(val_split * n)

    train_idx = idx[:n_train]
    valid_idx = idx[n_train:n_train + n_valid]
    test_idx  = idx[n_train + n_valid:]

    x_train, y_train = x[train_idx], y[train_idx]
    x_valid, y_valid = x[valid_idx], y[valid_idx]
    x_test,  y_test  = x[test_idx],  y[test_idx]

    return x_train, x_valid, x_test, y_train, y_valid, y_test

# 2. use our own splitter: 60 / 20 / 20
x_train, x_valid, x_test, y_train, y_valid, y_test = data_split(X, y, train_split=0.6, val_split=0.2)


# 3. Build vocab (stoi) on TRAIN ONLY
stoi = get_stoi(x_train)

# 4. Tokenize + encode labels
x_train_tok, y_train_enc = tokenize(x_train, y_train, stoi)
x_valid_tok, y_valid_enc = tokenize(x_valid, y_valid, stoi)
x_test_tok,  y_test_enc  = tokenize(x_test,  y_test,  stoi)

# 5. Pad / truncate to length 500
seq_len = 500

x_train_pad = padding_(x_train_tok, seq_len)
x_valid_pad = padding_(x_valid_tok, seq_len)
x_test_pad  = padding_(x_test_tok,  seq_len)

print("Train padded shape:", x_train_pad.shape)
print("Valid padded shape:", x_valid_pad.shape)
print("Test padded shape:",  x_test_pad.shape)
````

````python
from torch.utils.data import Dataset, DataLoader

class MovieReviewDataset(Dataset):

  def __init__(self, reviews, targets, tokenizer, max_len):
    self.reviews = reviews
    self.targets = targets
    self.tokenizer = tokenizer
    self.max_len = max_len

  def __len__(self):
    return len(self.reviews)

  def __getitem__(self, item):
    review = str(self.reviews[item])
    #obtain numerical labels
    target = 1 if self.targets[item] == 'positive' else 0

    encoding = self.tokenizer.encode_plus(
      review,
      add_special_tokens=True,
      max_length=self.max_len,
      return_token_type_ids=False,
      pad_to_max_length=True,
      return_attention_mask=True,
      return_tensors='pt',
      truncation = True
    )

    return {
      'review_text': review,
      'input_ids': encoding['input_ids'].flatten(),
      'attention_mask': encoding['attention_mask'].flatten(),
      'targets': torch.tensor(target, dtype=torch.long)
    }
````
